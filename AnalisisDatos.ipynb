```python
import pandas as pd
```

```python
#Importar y leer el archivo CSV
df_energia = pd.read_csv("consumo_energia_iot.csv")
df_energia
```

```python
#Selección Unidimensional
serie_consumo = df_energia["Consumo_kWh"]
serie_consumo
```

```python
#proyeccion tabular
vista = df_energia[["Municipio", "Estrato", "Voltaje_Promedio"]]
vista
```

```python
#Consulta de Lectura Específica (loc)
fila_150 = df_energia.loc[150]
fila_150
```

```python
#Depuración de duplicados
df_energia = df_energia.drop_duplicates()
df_energia
```

```python
#Filtrado booleano simple
alto_consumo = df_energia[df_energia["Consumo_kWh"] > 250.0]
alto_consumo
```

```python
#Filtrado categórico
subset = df_energia[df_energia["Municipio"].isin(["Chinchiná", "Villamaría"])]
subset
```

```python
#Filtrado por intervalo numérico
voltaje_optimo = df_energia[df_energia["Voltaje_Promedio"].between(118.0, 122.0)]
voltaje_optimo
```

```python
#Consulta con query
comercial_mora = df_energia.query("Estrato == 'Comercial' and Estado_Pago == 'Mora'")
comercial_mora
```

```python
#Filtrado compuesto con texto
condicion = (df_energia["Tipo_Medidor"].str.startswith("B")) & (df_energia["Horas_Pico_kWh"] > 40.0)
resultado = df_energia[condicion]
resultado
```

```python
#Tratamiento de nulos
df_energia["Consumo_kWh"].fillna(df_energia["Consumo_kWh"].mean(), inplace=True)
df_energia["Voltaje_Promedio"].fillna(120.0, inplace=True)
df_energia
```

```python
#Transformación condicional
def nivel_tarifa(consumo):
    if consumo < 120:
        return "Bajo"
    elif consumo <= 250:
        return "Medio"
    else:
        return "Alto"

df_energia["Nivel_Consumo"] = df_energia["Consumo_kWh"].apply(nivel_tarifa)
df_energia[["Consumo_kWh", "Nivel_Consumo"]]
```

```python
#Renombrado y reordenamiento
df_energia.rename(columns={"Consumo_kWh": "Energia_kWh", "Tipo_Medidor": "Tecnologia"}, inplace=True)

df_energia = df_energia[["ID_Medidor", "Nivel_Consumo"] + [col for col in df_energia.columns if col not in ["ID_Medidor", "Nivel_Consumo"]]]

df_energia
```

```python
#Agregación por municipio
resumen_municipio = df_energia.groupby("Municipio").agg(
    consumo_promedio=("Energia_kWh", "mean"),
    horas_pico_total=("Horas_Pico_kWh", "sum")
)
resumen_municipio
```

```python
#Métricas de rango personalizadas
def rango(series):
    return series.max() - series.min()

resumen_estrato = df_energia.groupby("Estrato").agg(
    rango_consumo=("Energia_kWh", rango),
    rango_voltaje=("Voltaje_Promedio", rango)
)
resumen_estrato
```

```python
#Agrupamiento multinivel
resumen_multi = df_energia.groupby(["Municipio", "Estrato"]).agg(
    consumo_promedio=("Energia_kWh", "mean"),
    voltaje_promedio=("Voltaje_Promedio", "mean")
)
resumen_multi
```
